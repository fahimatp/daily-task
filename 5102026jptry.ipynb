```python
day=3

match day:
    case 1:
        print("mon")
    case 2:
        print("th")
    case 3:
        print("w")
    
```

```python
day=9

match day:
    case 1|2|3|4|5:
        print("wd")
    case 6|7:
        print("we")
    case _:
        print("invalid")

        
```

```python
day=2
month=3

match day:
    case 1 if month==1:
        print("mon in jan")
    case 2 if month==2:
        print("th in jan")
    case 3 if month==3:
        print("w in jan")
    case 4 if month==4:
        print("t in jan")
    case 5 if month==5:
        print("f in jan")

    case 1 if month==2:
        print("mon in feb")
    case 2 if month==2:
        print("th in feb")

    case 1 if month==3:
        print("mo in march")
    case 2 if month==3:
        print("th in march")
```
