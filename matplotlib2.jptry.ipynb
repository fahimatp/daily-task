```python
import numpy as np
```

```python
import matplotlib.pyplot as plt
```

```python
data=np.random.randn(1000)
plt.hist(data,bins=30)
plt.title('histogram')
plt.show()
```

```python
sizes=[30,25,20,25]
labels=["A","B","C","D"]
plt.pie(sizes,labels=labels,autopct='%.1f%%')
plt.title("pie chart")
plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot([1,2,3],[1,4,9])
plt.title("plot 1")
plt.subplot(1,2,2)
plt.bar(["A","B","C"],[5,7,3])
plt.title("plot 2")
plt.tight_layout()
plt.show()
```

```python
plt.subplot(2,2,1)
plt.plot([1,2,3],[1,4,9])
plt.title("plot 1")

plt.subplot(2,2,2)
plt.bar(["A","B","C"],[5,7,3])
plt.title("plot 2")

plt.subplot(2,2,3)
sizes=[30,25,20,25]
labels=["A","B","C","D"]
plt.pie(sizes,labels=labels,autopct='%.1f%%')
plt.subplot(2,2,4)
plt.plot([1,2,3],[1,4,9])
plt.title("plot 1")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.subplot(2,2,1)
plt.plot([1,2,3],[1,4,9],label="sales")
plt.title("plot 1")
plt.grid()
plt.legend()

plt.subplot(2,2,2)
plt.bar(["A","B","C"],[5,7,3],label="production")
plt.title("plot 2")
plt.grid()
plt.legend()

plt.subplot(2,2,3)
sizes=[30,25,20,25]
labels=["A","B","C","D"]
plt.pie(sizes,labels=labels,autopct='%.1f%%')

plt.subplot(2,2,4)
plt.plot([1,2,3],[1,4,9])
plt.title("plot 1")
plt.grid()
plt.tight_layout()

plt.show()
```

```python
plt.plot([1,2,3],[1,4,9],label="sales")
plt.xticks([1,2,3,4,5,])
plt.yticks([1,4,9,8,7])
plt.title("plot 1")
plt.grid()
plt.legend()
plt.savefig("plot.png")
plt.show()
```

```python
data=[5,7,8,9,15,22,25,28]
plt.boxplot(data)
plt.title("box plot")
plt.show()
```

```python
days=[1,2,3,4,5]
sales=[100,200,150,300,400]
plt.fill_between(days,sales,color="purple")
plt.title("area plot")
plt.show()
```

```python
months=["jan","feb","mar","apr","may","jun"]

electronics=[120,150,180,170,210,240]
clothing=[80,90,100,120,130,150]
furniture=[50,60,70,65,80,90]
plt.stackplot(
months,
electronics,
clothing,
furniture,
labels=["electronics","clothing","furniture"]
)

plt.xlabel("month")
plt.ylabel("sales(thousands)")
plt.title("monthly sales by product category")

plt.legend()
plt.show()
```

```python
plt.plot(months,clothing)
plt.text("mar",130,"important point")
plt.show()
```
